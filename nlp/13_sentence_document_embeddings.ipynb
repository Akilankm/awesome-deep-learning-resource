## Change one thing at a time

| Change | Immediate effect | Downstream consequence |
|---|---|---|
| Change **pooling/aggregation** | different token evidence is emphasized | document vector geometry moves |
| Increase dimensionality reduction strength | vector becomes more compressed | speed improves but fine distinctions can vanish |
| Normalize vectors | magnitude is removed | cosine similarity becomes the natural comparison |

> Before changing a parameter, state the expected direction of the downstream effect.

### When to use

### When not to use / caution
Avoid relying only on a single vector when token-level evidence or exact spans matter.

### Debugging lens

```python
from pathlib import Path
import re, math, json, random, statistics
import numpy as np
import pandas as pd
np.random.seed(42)
random.seed(42)
pd.set_option('display.max_colwidth', 120)
DATA = Path('data')
print('Reproducibility seed: 42')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer
texts=[
 'reset my forgotten password','cannot sign into my account','charged twice on my card',
 'duplicate payment on statement','change profile email address','update account contact details']
vec=TfidfVectorizer(ngram_range=(1,2))
X=vec.fit_transform(texts)
svd=TruncatedSVD(n_components=4,random_state=42)
Z=Normalizer().fit_transform(svd.fit_transform(X))
from sklearn.metrics.pairwise import cosine_similarity
sim=cosine_similarity(Z)
pd.DataFrame(np.round(sim,2),index=texts,columns=[f'd{i}' for i in range(len(texts))])
```

`TF-IDF → SVD` is a classical dense document embedding baseline. Modern sentence-transformer embeddings are usually stronger for semantic similarity, but this baseline is transparent, fast, and fully offline.
