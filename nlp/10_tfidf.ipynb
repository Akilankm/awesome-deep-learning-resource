## Mental model

**term counts + corpus rarity → TF-IDF weighting → sparse evidence vector → linear model**

Focus on the transformation of information from left to right. Ask what representation changes before asking which library call implements it.

## Change one thing at a time

| Change | Immediate effect | Downstream consequence |
|---|---|---|
| Increase **term frequency** in one document | TF rises | that term contributes more to this document vector |
| Increase **document frequency** across corpus | IDF falls | the term becomes less discriminative |
| Enable vector normalization | vector magnitude is controlled | cosine/linear comparisons depend more on direction than length |

> Before changing a parameter, state the expected direction of the downstream effect.

### Debugging lens
Inspect TF, DF and final weight separately; a surprising TF-IDF value always comes from one of those components.

```python
from pathlib import Path
import re, math, json, random, statistics
import numpy as np
import pandas as pd
np.random.seed(42)
random.seed(42)
pd.set_option('display.max_colwidth', 120)
DATA = Path('data')
print('Reproducibility seed: 42')
```

For term $t$ in document $d$:

$$\mathrm{tfidf}(t,d)=\mathrm{tf}(t,d)\times\log\frac{N}{\mathrm{df}(t)}$$

Libraries often use smoothing and vector normalization. The key intuition is stable: **a term gets more weight when it is frequent in this document but uncommon across the corpus**.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
docs=['refund delayed refund','refund completed','login delayed']
vec=TfidfVectorizer(norm=None,smooth_idf=False)
X=vec.fit_transform(docs)
pd.DataFrame(np.round(X.toarray(),3),columns=vec.get_feature_names_out())
```

```python
terms=vec.get_feature_names_out()
idf=dict(zip(terms,np.round(vec.idf_,3)))
print('IDF:',idf)
print('Most discriminative in doc1:', terms[X.toarray()[0].argmax()])
```

```python

import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(7,3))
im=ax.imshow(X.toarray(),aspect='auto')
ax.set_xticks(range(len(terms))); ax.set_xticklabels(terms,rotation=45,ha='right')
ax.set_yticks(range(len(docs))); ax.set_yticklabels([f'doc{i+1}' for i in range(len(docs))])
ax.set_title('TF-IDF document–term matrix')
fig.colorbar(im,ax=ax,label='weight')
plt.tight_layout(); plt.show()
```
