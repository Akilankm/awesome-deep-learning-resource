# 07 — From Scratch to PyTorch

Once the mechanics are understood, PyTorch becomes a productivity layer rather than magic. It automates tensor bookkeeping, computational graphs, differentiation, and parameter updates while preserving the same mathematics.

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
torch.manual_seed(42)
```

## 1. Autograd is automatic backpropagation

```python
x=torch.tensor(2.0)
w=torch.tensor(-0.5,requires_grad=True)
b=torch.tensor(0.0,requires_grad=True)
y=torch.tensor(5.0)
yhat=w*x+b
loss=(yhat-y)**2
loss.backward()
print('loss =',loss.item())
print('dL/dw =',w.grad.item())
print('dL/db =',b.grad.item())
```

Compare these gradients with Notebook 04: PyTorch is computing the same chain-rule result automatically.

## 2. Train XOR with `nn.Module`

```python
X=torch.tensor([[0.,0.],[0.,1.],[1.,0.],[1.,1.]])
y=torch.tensor([[0.],[1.],[1.],[0.]])
class XORNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.net=nn.Sequential(nn.Linear(2,4),nn.Tanh(),nn.Linear(4,1))
    def forward(self,x): return self.net(x)
model=XORNet()
criterion=nn.BCEWithLogitsLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=.05)
losses=[]
for epoch in range(1500):
    optimizer.zero_grad(); logits=model(X); loss=criterion(logits,y); loss.backward(); optimizer.step(); losses.append(loss.item())
with torch.no_grad(): probs=torch.sigmoid(model(X))
print(probs)
```

## 3. Training curve

```python
plt.figure(figsize=(9,4)); plt.plot(losses); plt.yscale('log'); plt.xlabel('epoch'); plt.ylabel('loss'); plt.title('PyTorch training curve'); plt.grid(alpha=.25); plt.show()
```

## 4. Inspect learned parameters

```python
for name,param in model.named_parameters():
    print(name,param.shape)
    print(param.detach().numpy())
```

Framework abstraction should come **after** first-principles understanding. The conceptual loop is still forward → loss → backward → update.
