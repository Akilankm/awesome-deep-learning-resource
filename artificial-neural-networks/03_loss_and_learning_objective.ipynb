# 03 — Loss Functions: How a Network Knows It Is Wrong

A neural network does not understand mistakes directly; it receives a scalar **loss**.

Regression often uses mean squared error, while binary classification commonly uses binary cross-entropy. Training adjusts parameters to reduce this loss.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Visualize squared error

```python
target=3.0
preds=np.linspace(-2,8,300)
loss=(preds-target)**2
plt.figure(figsize=(9,5)); plt.plot(preds,loss); plt.axvline(target,linestyle='--'); plt.scatter([target],[0],s=80); plt.xlabel('prediction'); plt.ylabel('squared error'); plt.title('Loss landscape for one prediction'); plt.grid(alpha=.25); plt.show()
```

## 2. One neuron creates a 3D loss surface

```python
x=2.0; y=5.0
ws=np.linspace(-1,4,150); bs=np.linspace(-3,3,150)
W,B=np.meshgrid(ws,bs); YHAT=W*x+B; LOSS=(YHAT-y)**2
fig=plt.figure(figsize=(9,7)); ax=fig.add_subplot(111,projection='3d'); ax.plot_surface(W,B,LOSS,alpha=.75); ax.set_xlabel('weight'); ax.set_ylabel('bias'); ax.set_zlabel('loss'); ax.set_title('Loss surface: parameters → error'); plt.show()
```

Every point on the surface is a possible model. Optimization asks which direction in parameter space decreases loss fastest.

## 3. Binary cross-entropy intuition

```python
p=np.linspace(.001,.999,400)
loss_y1=-np.log(p); loss_y0=-np.log(1-p)
plt.figure(figsize=(9,5)); plt.plot(p,loss_y1,label='true label = 1'); plt.plot(p,loss_y0,label='true label = 0'); plt.xlabel('predicted probability P(y=1)'); plt.ylabel('binary cross-entropy'); plt.ylim(0,7); plt.legend(); plt.grid(alpha=.25); plt.show()
```

A confident wrong prediction receives a large penalty. The loss surface is therefore the landscape that learning navigates.
