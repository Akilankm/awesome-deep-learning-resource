At initialization, hidden coordinates are determined by random weights. Backpropagation changes those weights. Therefore the **same image gets different hidden coordinates as training proceeds**.

## 3. Why this happens

For hidden neuron $j$: $h_j = \tanh(x^T w_j + b_j)$. Backpropagation computes $\partial L/\partial w_j$. The update $w_j \leftarrow w_j - \eta \partial L/\partial w_j$ changes the projection direction. On the next forward pass, $x^T w_j$ changes, so the point moves in hidden space.

This is **representation learning as geometry**, not a slogan.
