# 02 — Feed-Forward Neural Networks

A feed-forward neural network composes neurons into layers. Information moves only forward:

$$x \rightarrow h^{(1)} \rightarrow h^{(2)} \rightarrow \hat y$$

For a dense layer: $z^{(l)}=W^{(l)}a^{(l-1)}+b^{(l)}$ and $a^{(l)}=\phi(z^{(l)})$.

This notebook makes every matrix multiplication and hidden representation visible.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
np.set_printoptions(precision=4, suppress=True)
```

## 1. Build a tiny 2 → 3 → 1 network by hand

```python
x=np.array([[1.5],[-2.0]])
W1=np.array([[0.5,-0.4],[1.0,0.3],[-0.8,0.9]])
b1=np.array([[0.1],[-0.2],[0.3]])
W2=np.array([[1.2,-0.7,0.5]])
b2=np.array([[0.05]])
relu=lambda z: np.maximum(0,z)
sigmoid=lambda z: 1/(1+np.exp(-z))
z1=W1@x+b1
a1=relu(z1)
z2=W2@a1+b2
yhat=sigmoid(z2)
print('x =',x.ravel())
print('z1 =',z1.ravel())
print('a1 =',a1.ravel())
print('z2 =',z2.ravel())
print('prediction =',yhat.ravel())
```

## 2. Visualize activation flow

```python
def draw_network(values_in,values_h,value_out):
    fig,ax=plt.subplots(figsize=(10,5))
    xin=[(0,1.8),(0,.2)]; hid=[(1.5,2.4),(1.5,1),(1.5,-.4)]; out=(3,1)
    for i,p in enumerate(xin):
        ax.scatter(*p,s=1200); ax.text(*p,f'x{i+1}\n{values_in[i]:.2f}',ha='center',va='center')
    for j,p in enumerate(hid):
        ax.scatter(*p,s=1200); ax.text(*p,f'h{j+1}\n{values_h[j]:.2f}',ha='center',va='center')
    ax.scatter(*out,s=1200); ax.text(*out,f'ŷ\n{value_out:.3f}',ha='center',va='center')
    for p1 in xin:
        for p2 in hid: ax.plot([p1[0],p2[0]],[p1[1],p2[1]],alpha=.4)
    for p1 in hid: ax.plot([p1[0],out[0]],[p1[1],out[1]],alpha=.4)
    ax.axis('off'); ax.set_title('Forward propagation: values move left → right'); plt.show()
draw_network(x.ravel(),a1.ravel(),float(yhat))
```

The hidden layer is a **new representation of the input**, not merely an intermediate calculation.

## 3. Representation learning: watch coordinates change

```python
np.random.seed(7)
X=np.random.uniform(-2,2,size=(250,2))
y=((X[:,0]*X[:,1])>0).astype(int)
def hidden_representation(scale):
    W=np.array([[1.,1.],[1.,-1.]])*scale
    H=np.tanh(X@W.T)
    fig=plt.figure(figsize=(12,5))
    ax1=fig.add_subplot(121); ax1.scatter(X[:,0],X[:,1],c=y); ax1.set_title('Original input space'); ax1.set_xlabel('x1'); ax1.set_ylabel('x2')
    ax2=fig.add_subplot(122); ax2.scatter(H[:,0],H[:,1],c=y); ax2.set_title('Hidden representation'); ax2.set_xlabel('h1'); ax2.set_ylabel('h2')
    plt.show()
widgets.interact(hidden_representation,scale=widgets.FloatSlider(value=1,min=.1,max=4,step=.1));
```

## 4. Why nonlinear depth matters

Purely linear layers collapse into one linear transformation. Nonlinear activations prevent that collapse, allowing depth to repeatedly reshape representation space.
