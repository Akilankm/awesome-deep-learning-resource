# 04 — Gradients and Gradient Descent

A gradient answers: **if I change a parameter slightly, how will the loss change?**

For parameter $w$, gradient descent uses $w \leftarrow w-\eta \frac{\partial L}{\partial w}$.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import ipywidgets as widgets
```

## 1. Derive a gradient for one neuron

```python
x=2.0; y=5.0; w=-0.5; b=0.0
yhat=w*x+b
loss=(yhat-y)**2
dL_dyhat=2*(yhat-y)
dL_dw=dL_dyhat*x
dL_db=dL_dyhat
print('prediction:',yhat)
print('loss:',loss)
print('dL/dw:',dL_dw)
print('dL/db:',dL_db)
```

## 2. Animate gradient descent on the loss curve

```python
x=2.; y=5.; b=0.; lr=.08; w=-1.
history=[]
for step in range(25):
    yhat=w*x+b; loss=(yhat-y)**2; grad=2*(yhat-y)*x
    history.append((w,loss)); w-=lr*grad
grid=np.linspace(-2,4,300); curve=(grid*x+b-y)**2
fig,ax=plt.subplots(figsize=(9,5)); ax.plot(grid,curve); point,=ax.plot([],[],'o',markersize=10); title=ax.set_title(''); ax.set_xlabel('weight'); ax.set_ylabel('loss'); ax.grid(alpha=.25)
def animate(i):
    wi,li=history[i]; point.set_data([wi],[li]); title.set_text(f'step {i}: w={wi:.3f}, loss={li:.3f}'); return point,title
anim=FuncAnimation(fig,animate,frames=len(history),interval=350,blit=False); plt.close(fig); HTML(anim.to_jshtml())
```

The gradient is local information, but repeated updates trace a path downhill through parameter space.

## 3. Learning-rate experiment

```python
def run_descent(lr):
    w=-1.; losses=[]
    for _ in range(30):
        yhat=w*2; loss=(yhat-5)**2; grad=2*(yhat-5)*2; losses.append(loss); w-=lr*grad
    plt.figure(figsize=(8,4)); plt.plot(losses,marker='o'); plt.yscale('log'); plt.xlabel('step'); plt.ylabel('loss'); plt.title(f'learning rate={lr:.4f}'); plt.grid(alpha=.25); plt.show()
widgets.interact(run_descent,lr=widgets.FloatLogSlider(value=.05,base=10,min=-3,max=0,step=.1));
```
