# 10 — Real Data Walkthrough: Handwritten Digits

This notebook anchors ANN mechanics in **real handwritten digit data**. Every number comes from an actual 8×8 image.

![Real digit features](visual_learning/real_digit_features.svg)


## 1. One real sample becomes features

We choose dataset sample **index 3**, whose true label is **3**. Instead of jumping immediately to 64 anonymous pixels, we first derive four interpretable measurements from the image: top-half ink, bottom-half ink, left-half ink, and right-half ink.


```python
from sklearn.datasets import load_digits
import numpy as np
digits = load_digits()
img = digits.images[3]
label = int(digits.target[3])
features = np.array([img[:4,:].sum(), img[4:,:].sum(), img[:,:4].sum(), img[:,4:].sum()], dtype=float)
print("chosen label:", label)
for name, value in zip(["top_ink","bottom_ink","left_ink","right_ink"], features): print(f"{name} = {value:.2f}")
```

## 2. One neuron: contribution accounting

For $z = w^T x + b$, each term $w_i x_i$ has a visible contribution. This is the antidote to treating the neuron as a black box.

![Neuron contribution accounting](visual_learning/neuron_contributions_real_digit.svg)


```python
w = np.array([0.03,-0.02,0.04,-0.01]); b=-1.2
contributions = w*features
z = contributions.sum()+b
a = 1/(1+np.exp(-z))
print("contributions:", contributions)
print("z:", z)
print("sigmoid(z):", a)
```

## 3. What affects what?

- Increase an input while its weight is positive → $z$ rises.
- Increase an input while its weight is negative → $z$ falls.
- Increase bias → the whole neuron response shifts.
- Increase weight magnitude → that feature has more leverage.
